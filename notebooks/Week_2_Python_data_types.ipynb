## For loops

In programming, loops are a way to repeat code for each item in a list. They're useful because they can mean less repeated code and are a way to automate repetitive tasks.

They work with any sequence: lists, strings, ranges.

A `for` loop is a type of loop, and the one we will use the most commonly. 

Here is a really simple example that just prints out each item in the list:

````python
numbers = [1, 2, 3, 4, 5]

for n in numbers:
    print(n)
````

Something slightly more useful might be to double of each item in the list:

````python
numbers = [1, 2, 3, 4, 5]

for n in numbers:
    print(n * 2)
````

The list that we loop over can contain any kind of data, including strings:

````python
fruits = ["apple", "banana", "cherry"]

for f in fruits:
    print("I like", f)
````

They work on other sequences too. For example a for loop can iterate over all of the characters in a string:

````python
for letter in "Hello World":
    print(letter)
````

We can do more complex arithmetic in our loops:

````python
numbers = [2, 4, 6, 8]

for n in numbers:
    squared = n ** 2
    print(n, "squared is", squared)
````

We can also do things like keep a total of the values that we loop over:

````python
numbers = [3, 6, 9, 12]
total = 0

for n in numbers:
    total = total + n   # or total += n
    #print(total)

print("The total is", total)
````

### List comprehension

A 'list comprehension' is a compact way to create a new list that is based on an existing list.

For example lets say we want a new list that has all of the numbers in our first list doubled.

Using the 'normal' for loop, we could do that like this:

````python
numbers = [1, 2, 3, 4, 5]
doubled = []  # This makes a new, empty list

for n in numbers:
    # Loop over each n, double it, and add it to the end of the new list
    doubled.append(n * 2)  # The 'append' function adds a new number to our list
    
print(doubled)
````

That's fine, but a more compact way to do that would be:

````python
numbers = [1, 2, 3, 4, 5]

doubled = [n * 2 for n in numbers]

print(doubled)   # [2, 4, 6, 8, 10]
````

Commonly you will see letters like `n`, `x`, `i` or `j` to refer to elements in a list. However this is just convention and you could use anything you wanted to refer to the elements in the list e.g. `elephants`.

(Though typing `i` is a lot quicker!)

````python
# multiply element by 3, repeat for every element in the list 'numbers'
[elephants*3 for elephants in numbers]
````

## Data Types

### Data types recap

As we covered in the lecture there are different types of data: 

<b>Objects:</b> also known as strings or written characters/text in plain english e.g. Hello World

<b>Integers:</b> Whole numbers e.g. 2, 57 or 109567835

<b>Floats:</b> A number with a decimal place e.g. 2.34534, 5.5 or 1.0

<b>Boolean:</b> True or False data type
<br>

<b>Datetime:</b> Values that are either a date, time or both e.g. 2019-10-31 09:26:03.478039 (9:26 am on Halloween 2019)

<b>Category:</b> A finite list of text values E.g. London, Paris, Berlin, Rome (There are a finite number of capital cities)
