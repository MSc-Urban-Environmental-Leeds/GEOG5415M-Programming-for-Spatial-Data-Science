# Week 10: GitHub and Collaborative Research

In this practical you will gain experience of working collaboratively using GitHub. You will see how to fork a repository (make a copy of it in your own GitHub space), make changes, and then merge your changes back into the parent repository. The whole class will work on the same repository, so you will be able to see everyone's contributions to the same codebase.

Specifically, you will do the following:

 - _fork_ a class repository (i.e. make a copy of the repository in your GitHub account)
 - open a file from that repository, `points.py`, using GitHub
 - edit that file, using GitHub, to add a new data point for a location that is important to you
 - _commit_ your changes to `points.py` to your repository.
 - open the file `points_map.ipynb` in Colab to check that you edited `points.py` correctly
 - create a _pull request_ to ask the owner of the class repository (Nick) to merge your changes into the main copy
 - have a look at the main copy to see if your point has been added to the map




## Fork the class repository

We have a shared repository for this class practical, which you can find at: https://github.com/MSc-Urban-Environmental-Leeds/GEOG5415M-Class_Repository

**Go to that repository now and have a look at the files**. There are two files that are important:

  - `points.py`: this is a Python script that contains a list of data points (latitude and longitude coordinates) for locations that are important to the students in this class. Each student is going to edit this file independently, adding some points that are important to them.

## Create a pull request to merge your changes

Let's recap what we've done so far:

  - Forked the class repository (called 'GEOG5415M-Class_Repository') to make a copy in our own GitHub account. You 'own' this copy, but it remains linked to the parent repository so we can keep them synchronised.
  - Edited the file `points.py` in your forked repository to added two points
  - Committed those changes to your forked repository. Now _your_ copy of the repository contains the new points, but the parent repository does not.
  - Opened your copy of the notebook `points_map.ipynb` in Colab and ran it to check that our points were added correctly

The final stage is to ask the owner of the parent repository (Nick & Fran) to merge your changes into the main copy. This is done using a _pull request_.

In your repository, **click on the 'Pull requests' tab**, and then **click the green 'New pull request' button**.

![Picture of the new pull request button](https://raw.githubusercontent.com/MSc-Urban-Environmental-Leeds/GEOG5415M-Programming-for-Spatial-Data-Science/main/notebooks/screenshots/week10-create_pull_request.png)

Once you have created the pull request, you will be taken to a new page where you can see the changes that you have made. Lines in green show additions, and lines in red show deletions. In my example below, I have added two points (shown in green) and deleted those two commented lines (shown in red).

![Picture of the pull request diff view](https://raw.githubusercontent.com/MSc-Urban-Environmental-Leeds/GEOG5415M-Programming-for-Spatial-Data-Science/main/notebooks/screenshots/week10-pull_request.png)

**Click on green 'Create pull request' button** to finalise the pull request. You will be asked to leave a message - write something sensible, e.g. "Added two points for locations important to me".

Now **come and see Nick or Fran** to let them know that you have created a pull request. They will review your changes and, if everything looks good, they will merge your changes into the parent repository. Then you will be able to see your points in the main class map.

And that's it! You have successfully used GitHub to work collaboratively on a codebase.
