# Week 10: GitHub and Collaborative Research

In this practical you will gain experience of working collaboratively using GitHub. You will see how to fork a copy, make changes, and then merge your changes back into the parent repository. The whole class will work on the same repository, so you will be able to see everyone's contributions to the same codebase.

Specifically, you will do the following:

 - _fork_ a class repository (i.e. make a copy of the repository in your GitHub account)
 - open a file from that repository, `points.py`, in Colab
 - edit that file to add a new data point for a location that is important to you
 - open the file `points_map.ipynb` in Colab to check that you edited `points.py` correctly
 - _commit_ your changes to `points.py`
 - _push_ your local changes (in Colab) to your GitHub repository
 - create a _pull request_ to ask the owner of the class repository to merge your changes into the main copy
 - have a look at the main copy to see if your point has been added to the map
